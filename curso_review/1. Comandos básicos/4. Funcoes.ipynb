````python
# Exemplo 1

def func_01(a):
    """
    Esse é o documentation string (docstring) da função func_01.
    Essa função simplesmente imprime o valor do argumento passado por parâmetro.
    """
    print(a)
````

````python
def inc(num):
    """"
    Essa função recebe um número como argumento e retorna o número incrementado em 1.
    """
    return num +1
````

````python
inc(2)
````

````python
a = inc(5)
a
````

````python
b = inc(10)
b
````

````python
def raiz(r, num):
    """
    Essa função recebe dois argumentos: r (raiz) e num (número).
    Ela retorna a raiz r-ésima do número num.
    """
    return num ** (1/r)
````

````python
a = raiz(2,9)
a
````

````python
d= raiz(r = 2, num = 9)
d
````

````python
# Exemplo 3 - parâmetros default
def raiz(num, r=2):
    """
    Essa função recebe dois argumentos: num (número) e r (raiz).
    O argumento r tem um valor default de 2, o que significa que se não for fornecido, será considerado como 2.
    Ela retorna a raiz r-ésima do número num.
    """
    return num ** (1/r)
````

````python
a = raiz(16)
a
````

````python
# Exemplo 04 Utilizando args

def soma(*nums):
    """
    Essa função recebe um número variável de argumentos e retorna a soma de todos eles.
    """
    print(type(nums))
    soma = 0
    for num in nums:
        soma += num # é igual a soma = soma + num

    return soma
````

````python
soma(2, 3, 4, 5, 6, 7, 8, 9)
````
