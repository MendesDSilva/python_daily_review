```python
# 4. Crie uma condição que classifique um pedido como concluído, pendente ou cancelados. (UTILIZANDO MATCH-CASE)

id_pedido = 25896
nome = 'Produto A'
preco = 105.60
quantidade = 50
subtotal = preco * quantidade
status = 'cancelado'
#print(f"subtotal: R${subtotal:.2f}")

match status:
    case 'concluído':
        print(f"O pedido: {id_pedido} do produto {nome} e preco R${preco:.2f} de quantidade {quantidade} foi CONCLUÍDO com SUCESSO, com subtotal de R${subtotal:.2f}")

    case 'pendente':
         print(f"O pedido: {id_pedido} do produto {nome} e preco R${preco:.2f} de quantidade {quantidade} ainda está PENDENTE, com subtotal de R${subtotal:.2f}")

    case 'cancelado':
        print(f"O pedido: {id_pedido} do produto {nome} e preco R${preco:.2f} de quantidade {quantidade} foi CANCELADO, com subtotal de R${subtotal:.2f}")

    case _:
        print(f"Erro: status '{status}' não reconhecido.")
```
