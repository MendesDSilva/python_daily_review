````python
# 4. Crie uma condicao que classifique um pedido como concluido, pendente ou cancelados. (UTILIZANDO IF)

id_pedido = 25896
nome = 'Produto A'
preco = 105.60
quantidade = 50
subtotal = preco * quantidade
status = 'cancelado'
#print(f"subtotal: R${subtotal:.2f}")

if status == 'concluido':
    print(f"O pedido: {id_pedido} do produto {nome} e preco R${preco:.2f} de quantidade {quantidade} foi CONCLUÍDO com SUCESSO, com subtotal de R${subtotal:.2f}")

elif status == 'pendente':
    print(f"O pedido: {id_pedido} do produto {nome} e preco R${preco:.2f} de quantidade {quantidade} ainda está PENDENTE, com subtotal de R${subtotal:.2f}")

elif status == 'cancelado':
    print(f"O pedido: {id_pedido} do produto {nome} e preco R${preco:.2f} de quantidade {quantidade} foi CANCELADO, com subtotal de R${subtotal:.2f}")

else:
    print(f"Erro: status '{status}' não reconhecido.")
````

````python
# 4. Crie uma condicao que classifique um pedido como concluido, pendente ou cancelados. (UTILIZANDO MATCH-CASE)

id_pedido = 25896
nome = 'Produto A'
preco = 105.60
quantidade = 50
subtotal = preco * quantidade
status = 'cancelado'
#print(f"subtotal: R${subtotal:.2f}")

match status:
    case 'concluído':
        print(f"O pedido: {id_pedido} do produto {nome} e preco R${preco:.2f} de quantidade {quantidade} foi CONCLUÍDO com SUCESSO, com subtotal de R${subtotal:.2f}")

    case 'pendente':
         print(f"O pedido: {id_pedido} do produto {nome} e preco R${preco:.2f} de quantidade {quantidade} ainda está PENDENTE, com subtotal de R${subtotal:.2f}")

    case 'cancelado':
        print(f"O pedido: {id_pedido} do produto {nome} e preco R${preco:.2f} de quantidade {quantidade} foi CANCELADO, com subtotal de R${subtotal:.2f}")

    case _:
        print(f"Erro: status '{status}' não reconhecido.")
````

````python
# 5. Explique a diferença entre == e is com um exemplo prático.

# Em Python, os operadores == e is são utilizados para realizar comparações, mas possuem finalidades diferentes.
# - == (igualdade): verifica se dois objetos possuem o mesmo valor.
# - is (identidade): verifica se duas variáveis apontam para o mesmo objeto na memória.

produto_a = 100
produto_b = 100

print(produto_a == produto_b) 
````

````python
# == retorna True porque as listas possuem os mesmos elementos. 
# - is retorna False porque são duas listas distintas na memória.

produto_a = [100, 200, 300]
produto_b = [100, 200, 300]

print(produto_a == produto_b) 
print(produto_a is produto_b)
````

````python
# 7. Utilize range() para gerar dez identificadores sequenciais.

# range(inicio, fim, passo)
# inicio - Primeiro numero da sequencia
# fim - Limite superior da sequencia (não incluso)
# passo - Incremento entre os numeros da sequencia

for ide in range(1,11,1):
    print(f"Identificador sequencial: {ide}")
````

````python
# 8. Explique a diferença entre break, continue e pass 
# break - Interrompe o loop e encerra a execução do mesmo.
# continue - Pula a interação atual do loop e passa para a próxima iteração.
# pass - É uma instrução nula que não faz nada. É usada como um placeholder em blocos de código onde uma declaração é sintaticamente necessária, mas você não deseja executar nenhuma ação.
````

````python
# 8.1 break - Executa até o critério e não traz ele.

for id in range(1,11,1):
    if id == 5:
        break
    print(f"Identificador sequencial: {id}")
````

````python
# 8.2 continue - Pula o critério do loop e passa para a próxima iteração.

for id in range(1,11,1):
    if id == 5:
        continue
    print(f"Identificador sequencial: {id}")
````

````python
# 8.3 pass - Não faz nada, mas é usado como um placeholder.

for id in range(1,11,1):
    if id == 5:
        pass
    print(f"Identificador sequencial: {id}")
````

````python
# Exemplo aplicado à Engenharia de Dados

pedidos = [
    {"id": 101, "status": "concluído"},
    {"id": 102, "status": "cancelado"},
    {"id": 103, "status": "pendente"},
    {"id": 104, "status": "concluído"}
]

for pedido in pedidos:

    if pedido["status"] == "cancelado": 
        continue         # Pula pedidos cancelados                
    
    if pedido["status"] == "pendente":
        print(f"Pedido {pedido['id']} está pendente. Encerrando a verificação.") 
        break            # Interrompe o loop se encontrar um pedido pendente
    
    print(f"Processando pedido {pedido['id']}")
````

````python
# 9.1 - Exemplo aplicado à Engenharia de Dados - Calculando o faturamento de cada pedido

pedidos = [
    {"id": 101, "quantidade": 15,  "preco": 25.00},
    {"id": 102, "quantidade": 50,  "preco": 22.00},
    {"id": 103, "quantidade": 100, "preco": 23.00},
    {"id": 104, "quantidade": 200, "preco": 50.00}
]

def faturamento(pedidos):
    for ped in pedidos:
        subtotal = ped["quantidade"] * ped["preco"]
        
        print(f"O Faturamento do Pedido: {ped['id']} é R$ {subtotal:.2f}")

faturamento(pedidos)

````

````python
# 9.2 - Exemplo aplicado à Engenharia de Dados - Calculando o faturamento de cada pedido

pedidos = [
    {"id": 101, "quantidade": 15,  "preco": 25.00},
    {"id": 102, "quantidade": 50,  "preco": 22.00},
    {"id": 103, "quantidade": 100, "preco": 23.00},
    {"id": 104, "quantidade": 200, "preco": 50.00}
]


def faturamento_total(pedidos):
    total = 0

    for ped in pedidos:
        subtotal = ped["quantidade"] * ped["preco"]
        total += subtotal

    return total

resultado = faturamento_total(pedidos)

print(f"O faturamento total é: R$ {resultado:.2f}")


````

````python
# 12.1 - Exemplo aplicado à Engenharia de Dados 

pedidos = [
    {"id": "101", "quantidade": "15",  "preco": "25.00"},
    {"id": "102", "quantidade": "50",  "preco": "22.00"},
    {"id": "103", "quantidade": "100", "preco": "23.00"},
    {"id": "104", "quantidade": "200", "preco": "50.00"}
]


def conv_float(pedidos):

    for ped in pedidos:
        ped["id"] = int(ped["id"])
        ped["quantidade"] = int(ped["quantidade"])
        ped["preco"] = float(ped["preco"])

    return pedidos 

resultado = conv_float(pedidos)

print(resultado)
````

````python
# 15. Utilize zip() para associar listas de produtos e preços, tratando tamanhos diferentes.
# A função zip() permite percorrer duas ou mais listas simultaneamente, associando os elementos de acordo com suas posições.

produtos = ["Notebook", "Mouse", "Teclado"]
precos = [3500.00, 150.00, 250.00]

for produto, preco in zip(produtos,precos):
    print(f"Produto: {produto} | Preço: R$ {preco:.2f}")

````
