```python
# 1. Explique a diferença entre int, float, str e bool usando campos do e-commerce.

# int: Número inteiro sem casas decimais, Ex: 1045 , 500, 1
# float: Número decimal com casas decimais, Ex: 99.99, 49.95, 0.75
# str: Cadeia de caracteres, usada para representar texto, Ex: "Produto A", "Descrição do produto", "Categoria"
# bool: Valor booleano que representa verdadeiro ou falso, Ex: True (produto disponível), False (produto indisponível)
```
